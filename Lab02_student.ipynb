### E10 · Rows and columns – two ways to read the same table
**The idea:** you can look at a table in two ways – and you will need both.

**View 1 – by rows** (what `loader.py` gives you): a **list** of rows, each row a **dictionary**.
First choose the **row**, then the **column**:
```python
week_typed[3]["pm10"]      # row 4 (index 3) -> column pm10 -> 41
```

**View 2 – by columns:** a **dictionary** of columns, each column a **list**.
First choose the **column**, then the **position**:
```python
columns = {"temp_c": [11.5, 9.8, 8.1, 10.2, ...],
           "pm10":   [38,   22,  18,  41,   ...]}
columns["pm10"][3]         # column pm10 -> position 3 -> 41   (the same cell!)
```

| You want… | Use |
|---|---|
| everything about **one day** (one observation) | by rows: `week_typed[3]` |
| a statistic of **one variable** (min, max, mean) | by columns: `columns["pm10"]` |

⚠️ Positions start at **0**: the 4th row is `[3]`, the 5th value is `[4]`.
